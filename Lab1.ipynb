```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
 
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
 
GENRE_COLS = [
    "Action", "Adventure", "Avant Garde", "Boys Love", "Comedy", "Drama",
    "Fantasy", "Girls Love", "Gourmet", "Horror", "Josei", "Kids",
    "Mystery", "Romance", "Sci-Fi", "Seinen", "Shoujo", "Shounen",
    "Slice of Life", "Sports", "Supernatural", "Suspense",
]
 
 
def load_and_prepare(path: str) -> pd.DataFrame:
    df = pd.read_csv(path)
 
    # 1. Flag imputed rows BEFORE any modification
    df["anilist_score_imputed"] = df["anilist_score"].isna()
 
    # 2. Feature engineering
    df["log_mal_scored_by"] = np.log1p(df["mal_scored_by"])
    df["year_filled"] = df["year"].fillna(df["year"].median())  # small # missing years
 
    return df
 
 
def build_feature_matrix(df: pd.DataFrame) -> pd.DataFrame:
    """Numeric + one-hot features used to predict anilist_score."""
    type_dummies = pd.get_dummies(df["type"], prefix="type", drop_first=True)
    status_dummies = pd.get_dummies(df["status"], prefix="status", drop_first=True)
    genre_block = df[GENRE_COLS].astype(int)
 
    X = pd.concat(
        [
            df[["mal_score", "log_mal_scored_by", "year_filled", "episodes"]],
            type_dummies,
            status_dummies,
            genre_block,
        ],
        axis=1,
    )
    return X
 
 
def validate_models(X_obs: pd.DataFrame, y_obs: pd.Series):
    """Compare candidate imputation approaches with 5-fold CV on observed rows."""
    kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    scaler = StandardScaler()
 
    results = {"mean_baseline": [], "mal_score_only": [], "linear_multivariate": []}
 
    for train_idx, test_idx in kf.split(X_obs):
        X_train, X_test = X_obs.iloc[train_idx], X_obs.iloc[test_idx]
        y_train, y_test = y_obs.iloc[train_idx], y_obs.iloc[test_idx]
 
        # Baseline 1: mean imputation
        pred_mean = np.full(len(y_test), y_train.mean())
        results["mean_baseline"].append(mean_absolute_error(y_test, pred_mean))
 
        # Baseline 2: simple linear regression on mal_score alone
        lr_simple = LinearRegression().fit(X_train[["mal_score"]], y_train)
        pred_lr_simple = lr_simple.predict(X_test[["mal_score"]])
        results["mal_score_only"].append(mean_absolute_error(y_test, pred_lr_simple))
 
        # Candidate: multivariate Linear Regression
        X_train_s = scaler.fit_transform(X_train)
        X_test_s = scaler.transform(X_test)
        lr_multi = LinearRegression().fit(X_train_s, y_train)
        pred_lr_multi = lr_multi.predict(X_test_s)
        results["linear_multivariate"].append(mean_absolute_error(y_test, pred_lr_multi))
 
    print("5-fold CV Mean Absolute Error (lower is better):")
    for name, scores in results.items():
        print(f"  {name:20s}: {np.mean(scores):.4f} (+/- {np.std(scores):.4f})")
    print()
 
 
def fit_final_model(X_obs: pd.DataFrame, y_obs: pd.Series):
    """Fit the chosen model on ALL observed data, with a held-out check first."""
    X_train, X_test, y_train, y_test = train_test_split(
        X_obs, y_obs, test_size=0.2, random_state=RANDOM_STATE
    )
 
    scaler = StandardScaler().fit(X_train)
    lr = LinearRegression().fit(scaler.transform(X_train), y_train)
    preds = lr.predict(scaler.transform(X_test))
 
    print("Held-out test performance (20% split) of final Linear Regression model:")
    print(f"  MAE  : {mean_absolute_error(y_test, preds):.4f}")
    print(f"  RMSE : {mean_squared_error(y_test, preds) ** 0.5:.4f}")
    print(f"  R^2  : {r2_score(y_test, preds):.4f}\n")
 
    # Refit on ALL observed rows for the actual imputation
    scaler_full = StandardScaler().fit(X_obs)
    lr_full = LinearRegression().fit(scaler_full.transform(X_obs), y_obs)
 
    # Residuals from full model, used to add realistic noise back in (Step 5)
    fitted_full = lr_full.predict(scaler_full.transform(X_obs))
    residuals = y_obs.values - fitted_full
 
    return lr_full, scaler_full, residuals
 
 
def stochastic_impute(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    X = build_feature_matrix(df)
 
    obs_mask = ~df["anilist_score_imputed"]
    X_obs, y_obs = X[obs_mask], df.loc[obs_mask, "anilist_score"]
    X_missing = X[~obs_mask]
 
    validate_models(X_obs, y_obs)
    model, scaler, residuals = fit_final_model(X_obs, y_obs)
 
    if len(X_missing) > 0:
        point_preds = model.predict(scaler.transform(X_missing))
 
        # Stochastic component: sample noise from the observed residual
        # distribution so imputed values carry realistic variance instead
        # of sitting exactly on the regression line.
        noise = np.random.choice(residuals, size=len(point_preds), replace=True)
        imputed_values = point_preds + noise
 
        # Clip to the observed valid score range
        lo, hi = df["anilist_score"].min(), df["anilist_score"].max()
        imputed_values = np.clip(imputed_values, lo, hi)
 
        df.loc[~obs_mask, "anilist_score"] = imputed_values
 
    # Recompute the derived gap column for imputed rows
    df["score_gap_mal_anilist"] = df["anilist_score"] - df["mal_score"]
 
    return df
 
 
if __name__ == "__main__":
    INPUT_PATH = "animeoutput.csv"
    OUTPUT_PATH = "animeoutput_imputed.csv"
 
    df = load_and_prepare(INPUT_PATH)
    print(f"Rows: {len(df)} | Missing anilist_score: {df['anilist_score_imputed'].sum()}\n")
 
    df_imputed = stochastic_impute(df)
 
    n_imputed = df_imputed["anilist_score_imputed"].sum()
    print(f"Imputed {n_imputed} values.")
    print("\nSummary comparison (observed vs. imputed anilist_score):")
    print(
        df_imputed.groupby("anilist_score_imputed")["anilist_score"]
        .agg(["count", "mean", "std", "min", "max"])
    )
 
    df_imputed.drop(columns=["log_mal_scored_by", "year_filled"]).to_csv(
        OUTPUT_PATH, index=False
    )
    print(f"\nSaved imputed dataset to {OUTPUT_PATH}")

```

### Ranking by Genre

```python
# Grabbing the average rating for each genre
genre_avg_ratings = {}

for genre in genre_cols:
    genre_avg_ratings[genre] = df.loc[df[genre], 'mal_score'].mean()

genre_avg_ratings = pd.Series(genre_avg_ratings).sort_values(ascending=False)
#print(genre_avg_ratings) - Just for checking values

# Plotting the average ratings for each genre
genre_avg_ratings.plot(kind='bar', figsize=(12, 6))
plt.ylabel('Average MAL Score')
plt.title('Average MAL Rating by Genre')
plt.xticks(rotation=45, ha='right')
plt.ylim(5,7.5)
plt.tight_layout()
plt.show()

# Showing how many titles are in each genre
genre_counts = {genre: df[genre].sum() for genre in genre_cols}
genre_summary = pd.DataFrame({
    'avg_mal_score': genre_avg_ratings,
    'count': pd.Series(genre_counts)
}).sort_values('avg_mal_score', ascending=False)

print(genre_summary)
```

As seen above, the most popular demographic is Shonen, and the most popular genre is Suspense. However, it's important to take note of the number of shows that qualify for each category. Action shows are the most represented genre (having over 2000 entries) while Suspense only has 281. That means that each rating in Suspense has vastly more weight than Action does, which may contribute to Suspense's higher overall ranking. Kids show also has quite the low rating, but that may be because the target demographic (being children) are not the ones reviewing the shows; adult watchers are.
