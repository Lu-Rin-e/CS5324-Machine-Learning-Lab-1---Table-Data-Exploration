### Data Types
There are a lot of columns in this dataset that we don't need. Focusing on the columns we do, they can be classified as:
- Type: Categorical. There are two types: TV and MOVIE. All others have been filtered out
- Source: Categorical. This is what the anime is based on (e.g., Manga, Light Novel). 
- Age_Rating: Categorical. This is the broadcast age rating (e.g., G, PG, PG-13, R)
- Is_Adult: Categorical. This is whether the show is explicit or not (True/False). All Explicit shows will be filtered out, and this column will be deleted.
- Year: Numeric. The year the anime was released (e.g., 2026, 1980)
- Episodes: Numeric. The number of episodes a show has. All Movies will be listed as having 1 episode, and currently-airing shows/shows without a known episode count are listed as having 0 episodes
- Mal_Score: Ordinal. The score of the show on MAL (e.g., 5.6, 8.9)
- Mal_Scored_By: Numeric. The number of people who scored a show on MAL.
- Anilist_Score: Ordinal. The score of the show on Anilist (e.g., 5.6, 8.9)
- Score_Gap_Mal_Anilist: Numeric. The difference between the MAL score and Anilist score (e.g., -0.5)
- Genres: Categorical. The genre(s) and demographics of the show. Will be completed with multi-hot encoding, as some shows have multiple genres 
- Num_Platforms: Numerical. How many streaming platforms have the show available for viewers

```python
# Removing the columns that are not needed for the analysis (anime_id, name)
for col in ['mal_id', 'anilist_id', 'title_english', 'title_romaji', 'title_synonyms','title_japanese', 'season', 'decade', 'season_year', 'mal_rank', 'mal_popularity', 'mal_favorites', 'mal_members', 'anilist_score_100', 'anilist_favorites', 'anilist_popularity', 'studio_main', 'studios_all', 'top_tags', 'streaming_platforms', 'is_crunchyroll', 'is_netflix', 'is_hidive', 'is_funimation', 'is_amazon_prime', 'is_disney_plus', 'is_hulu', 'is_vrv', 'is_bilibili', 'is_youtube', 'score_tier', 'popularity_tier', 'synopsis', 'image_url', 'is_movie','duration_min','is_long_running']:
    if col in df:
        del df[col]

# Changing episodes data type to int - The reason it isn't like this originally is 
# because all currently-airing anime at the time have their number of episodes listed as "Unknown"
#df['episodes'] = pd.to_numeric(df['episodes'],errors='coerce')
#df['episodes'] = df['episodes'].fillna(0).astype(int)

# Seeing how much data is missing
import matplotlib
import matplotlib.pyplot as plt
import warnings
warnings.simplefilter('ignore', DeprecationWarning)
%matplotlib inline


import missingno as mn

mn.matrix(df.sort_values(by=['mal_score', 'anilist_score']))
plt.title("Sorted by MAL rating", fontsize=22)
plt.show()

#Seeing the types of data the dataset has
print(df.dtypes)
print('==========')
print(df.info())

```

```python
# Removing Adult/Explicit Anime - Most Western companies don't license explicit material
df = df[df.is_adult == False]

#Removing is_adult column
del df['is_adult']


#Multi-hot Encoding the Genres; they are currently one object
from sklearn.preprocessing import MultiLabelBinarizer

genre_lists = df['genres'].str.split('|').apply(lambda lst: [g.strip() for g in lst])

mlb = MultiLabelBinarizer()
genre_encoded = pd.DataFrame(
    mlb.fit_transform(genre_lists),
    columns=mlb.classes_,
    index=df.index
)


df.head()
```

```python



# The dataset combined MAL Themes, MAL Genres, and Demographics into genres. Just keeping the Genres and Demographics
genres_to_keep = [
    'Action', 'Adventure', 'Avant Garde', 'Boys Love',
    'Comedy', 'Drama', 'Fantasy', 'Girls Love', 'Gourmet', 'Horror',
    'Mystery', 'Romance', 'Sci-Fi', 'Slice of Life', 'Sports',
    'Supernatural', 'Suspense', 'Josei', 'Kids', 'Seinen', 'Shoujo', 'Shounen'
]

# Filtering the encoded columns before merging into dataframe
genre_encoded = genre_encoded[[col for col in genre_encoded.columns if col in genres_to_keep]]

df = pd.concat([df, genre_encoded], axis=1)


#Casting the columns to be booleans rather than integers
genre_cols = genre_encoded.columns  
df[genre_cols] = df[genre_cols].astype(bool)

#Deleting the old genres column

del df['genres']

# Removing all items that aren't TV shows or Movies

df = df[(df.type == 'TV') | (df.type == 'MOVIE')]


#Checking what the cleaned data looks like
df.to_csv('animeoutput.csv', index=False)

```
